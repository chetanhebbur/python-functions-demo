```python
# ============================================================
# math_utils.py - Enhanced Math Functions (v2)
# ============================================================

def add(a, b):
    """Returns sum of two numbers"""       # ← added docstring
    return a + b

def subtract(a, b):
    """Returns difference of two numbers""" # ← added docstring
    return a - b

def multiply(a, b):
    """Returns product of two numbers"""    # ← added docstring
    return a * b

def divide(a, b):
    """Returns division result"""           # ← added docstring
    if b == 0:                              # ← added zero division check
        raise ValueError("Cannot divide by zero!")
    return a / b

# 🆕 NEW FUNCTION
def power(a, b):
    """Returns a raised to the power of b"""
    return a ** b

# 🆕 NEW FUNCTION
def modulus(a, b):
    """Returns remainder of a divided by b"""
    return a % b
```

```python
# ============================================================
# string_utils.py - Enhanced String Functions (v2)
# ============================================================

def to_uppercase(text):
    """Converts text to uppercase"""        # ← added docstring
    if not isinstance(text, str):           # ← added type check
        raise TypeError("Input must be a string")
    return text.upper()

def to_lowercase(text):
    """Converts text to lowercase"""        # ← added docstring
    if not isinstance(text, str):           # ← added type check
        raise TypeError("Input must be a string")
    return text.lower()

def reverse_string(text):
    """Reverses a string"""                 # ← added docstring
    if not isinstance(text, str):           # ← added type check
        raise TypeError("Input must be a string")
    return text[::-1]

# 🆕 NEW FUNCTION
def count_words(text):
    """Returns number of words in a string"""
    return len(text.strip().split())

# 🆕 NEW FUNCTION
def is_palindrome(text):
    """Checks if a string is a palindrome"""
    cleaned = text.lower().replace(" ", "")
    return cleaned == cleaned[::-1]
```

```python
# ============================================================
# Test v2 - Enhanced functions
# ============================================================

print("=== Math Tests ===")
print(add(5, 3))           # 8
print(subtract(10, 4))     # 6
print(multiply(3, 4))      # 12
print(divide(10, 2))       # 5.0
print(power(2, 10))        # 🆕 1024
print(modulus(10, 3))      # 🆕 1

# Test divide by zero error handling
try:
    divide(10, 0)
except ValueError as e:
    print(f"Error caught: {e}")  # 🆕 Error caught: Cannot divide by zero!

print("\n=== String Tests ===")
print(to_uppercase("hello"))        # HELLO
print(to_lowercase("WORLD"))        # world
print(reverse_string("databricks")) # skcirbtaad
print(count_words("hello world"))   # 🆕 2
print(is_palindrome("racecar"))     # 🆕 True
print(is_palindrome("databricks"))  # 🆕 False
```
