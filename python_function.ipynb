```python
# ============================================================
# math_utils.py - Basic Math Functions (v1)
# ============================================================

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b
```

```python
# ============================================================
# string_utils.py - Basic String Functions (v1)
# ============================================================

def to_uppercase(text):
    return text.upper()

def to_lowercase(text):
    return text.lower()

def reverse_string(text):
    return text[::-1]
```

```python
# ============================================================
# Test v1 functions
# ============================================================

print("=== Math Tests ===")
print(add(5, 3))        # 8
print(subtract(10, 4))  # 6
print(multiply(3, 4))   # 12
print(divide(10, 2))    # 5.0

print("\n=== String Tests ===")
print(to_uppercase("hello"))       # HELLO
print(to_lowercase("WORLD"))       # world
print(reverse_string("databricks")) # skcirbtaad
```
